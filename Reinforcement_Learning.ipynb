```python
import numpy as np
import random
```

```python
start = (0,0)
goal = (4,4)
size = 5

actions = {
    "U" : (1,0),
    "D" : (-1,0),
    "L" : (0,-1),
    "R" : (0,1)
}
obstacles = [(1,2),(4,2)]
gamma = 0.8
alpha = 0.2
epsilon = 0.3
```

```python
def move(state, action):
    dr, dc = actions[action]
    r, c = state
    nxt = (dr + r,c + dc)

    if not(0<=nxt[0]<=size and 0<=nxt[1] <= size):
        return state
    if nxt in obstacles:
        return state
    return nxt
```

```python
for epoch in range(1000):
    state = start
    for i in range(100):
        Q.setdefault(state, {a : 0.0 for a in actions})
        if epsilon > random.random():
            action = random.choice(list(actions))
        else:
            action = max(Q[state], key = Q[state].get)
        next_state = move(state, action)

        if next_state == goal:
            reward = 10
        elif next_state == state:
            reward = -5
        else:
            reward = -1

        Q.setdefault(next_state, {a:0.0 for a in actions})
        old = Q[state][action]
        future = 0 if next_state == goal else max(Q[next_state].values())

        Q[state][action] = old + alpha*(reward + gamma*future - old)
        state = next_state
        if state == goal:
            break

state = start
path = [state]

for _ in range(20):
    if state == goal:
        break
    action = max(Q[state], key = Q[state].get)
    state = move(state,action)
    path.append(state)
print(path)
        
        
```

```python
grid = [["_" for _ in range(size)] for _ in range(size)]

for r, c in obstacles:
    grid[r][c] = "X"

for r, c in path:
    if grid[r][c] != "X":
        grid[r][c] == "*"

grid[start[0]][start[1]] = "S"
grid[goal[0]][goal[1]] = "G"

for row in grid:
    print(" ".join(row))
```
